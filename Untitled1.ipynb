```python
%%writefile app.py
import streamlit as st
import pandas as pd
from sklearn.ensemble import IsolationForest

st.title("💧 Water Fraud Detection System")

df = pd.read_csv("water_fraud_output.csv")

model = IsolationForest(contamination=0.05)
model.fit(df[['daily_usage', 'previous_avg']])

st.write("### 📊 Dataset")
st.dataframe(df)

fraud_data = df[df['fraud'] == -1]
st.write("### 🚨 Fraud Cases")
st.dataframe(fraud_data)

st.line_chart(df['daily_usage'])

st.write("## 🔍 Check New Usage")

usage = st.number_input("Enter Daily Usage", min_value=0)
avg = st.number_input("Enter Previous Avg", min_value=0)

if usage < 10 or avg < 10:
    st.warning("⚠️ Enter realistic values (above 10)")

if st.button("Check"):
    pred = model.predict([[usage, avg]])
    if pred[0] == -1:
        st.error("🚨 Fraud Detected")
    else:
        st.success("✅ Normal")
```
